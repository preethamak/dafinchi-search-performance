## 2. Audience distribution: who is searching, by observable signals?

### Device behavior

Bing desktop accounts for **96.7% of clicks** and about **89% of impressions**. Mobile accounts for **3.3% of clicks** and about **11% of impressions**. Mobile CTR is 0.59%, versus 2.13% desktop (3.6x lower), and its average position is also worse (7.39 vs 5.44).

If mobile had the desktop CTR on the current rounded 2.2K impressions, it would produce about **47 clicks** rather than 13, a roughly **34-click parity gap** for this period. This is a scenario, not a forecast; ranking and query mix differ by device.


```python
# Re-create the Bing device distribution chart from the source rows.
import matplotlib.pyplot as plt
bing_devices = [{"device":"Desktop","impressions":17800,"clicks":379}, {"device":"Mobile","impressions":2200,"clicks":13}]
click_total = 392
impression_total_rounded = 20000
labels = [r["device"] for r in bing_devices]
click_share = [100*r["clicks"]/click_total for r in bing_devices]
impression_share = [100*r["impressions"]/impression_total_rounded for r in bing_devices]
x = range(len(labels)); w=.34
fig, ax = plt.subplots(figsize=(8.3,3.7))
ax.bar([i-w/2 for i in x], impression_share, w, label="Share of impressions", color="#3268A8")
ax.bar([i+w/2 for i in x], click_share, w, label="Share of clicks", color="#2A8C82")
ax.set_xticks(list(x), labels); ax.set_ylim(0,105); ax.set_ylabel("Share of Bing total (%)")
ax.set_title("Bing search audience is strongly desktop-led", loc="left", weight="bold")
ax.grid(axis="y", alpha=.2); ax.legend(frameon=False, ncol=2); ax.set_axisbelow(True)
for bars in ax.containers: ax.bar_label(bars, fmt="%.1f%%", padding=2, fontsize=8)
for side in ["top","right"]: ax.spines[side].set_visible(False)
fig.tight_layout(); fig
```

The United States supplies **36.5% of Bing clicks** and about **70.5% of impressions**. China supplies **40.6% of clicks** from about **8.5% of impressions**, with a reported 9.29% CTR. That unusually strong click share is a good segment to inspect by query and landing page; it is not enough to conclude who those users are or whether localization is warranted.

## 4. Growth plan ranked by evidence

### Priority 1 - Restore useful Google coverage

Inspect a representative sample from the **18,916 discovered-not-indexed** and **6,925 crawled-not-indexed** groups. Classify URLs into valuable canonical pages, duplicate/parameter variants, intentionally excluded feeds, thin pages, and technical failures. Then correct the template or internal-link issue at the source. Do not submit all 26,810 URLs blindly; validate a small corrected set first.

### Priority 2 - Turn high-ranking impressions into clicks

For Whirlpool/Section 232, A. O. Smith, and Ameris, make the title and description state the exact company, quarter, and catalyst. Ensure the landing page immediately answers the query and clearly links to the source passage. Track CTR by query/page after the change.

### Priority 3 - Make the product answer the research job

On each catalyst page, present a compact brief: what changed, why it matters, numbers/guidance mentioned, and the transcript source. Add next steps: compare the company with peers, scan this call for a topic, or ask a question about this transcript. This connects acquisition pages to Dafinchi's comparison, Keyword Scan, and Earnings Chat experiences.

### Priority 4 - Fix mobile discovery and reading

Mobile has 11% of Bing impressions but only 3.3% of clicks; CTR is 3.6x below desktop and position is worse. Review mobile SERP titles/snippets and the first viewport of the highest-impression pages. Search Console has no Core Web Vitals data, so this dataset cannot diagnose page speed.

### Priority 5 - Learn from the segments before localizing

Inspect the query and landing pages behind China's 159 Bing clicks and 9.29% CTR, then compare them with the US. If the high CTR persists across relevant investor queries, consider a targeted country test. Do not infer language preference or user identity from country alone.

### Priority 6 - Measure the funnel

Google Analytics was inaccessible, and search reports do not expose sessions or product conversions. Track `organic landing -> transcript read -> Keyword Scan/Earnings Chat/comparison -> signup or return visit` in GA4 once the property is accessible.
