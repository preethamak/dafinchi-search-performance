## 3. Device distribution: where acquisition is strongest

Bing desktop drove **96.7% of clicks** from about **89% of impressions**. Mobile drove **3.3% of clicks** from about **11% of impressions**. Mobile CTR was **0.59%**, compared with **2.13%** on desktop; average position was also worse (7.39 vs. 5.44).

If mobile’s current rounded 2.2K impressions had desktop CTR, that would imply about **47 clicks**, versus 13 observed. The ~34-click difference is a scenario to size the gap, not a forecast: mobile query mix and rankings differ.

```python
import pandas as pd
import matplotlib.pyplot as plt

devices = pd.DataFrame([
    {"device":"Desktop", "impressions":17800, "clicks":379, "CTR (%)":2.13, "avg_position":5.44},
    {"device":"Mobile", "impressions":2200, "clicks":13, "CTR (%)":0.59, "avg_position":7.39},
])
devices["impression_share (%)"] = devices["impressions"] / 20000 * 100  # Bing total is rounded
devices["click_share (%)"] = devices["clicks"] / 392 * 100
devices[["device","impressions","clicks","CTR (%)","avg_position","impression_share (%)","click_share (%)"]]
```

```python
ax = devices.set_index("device")[["impression_share (%)","click_share (%)"]].plot(
    kind="bar", figsize=(8,4), color=["#3268A8","#2A8C82"]
)
ax.set_ylabel("Share of Bing total (%)")
ax.set_title("Bing search distribution by device")
ax.legend(["Impressions (approx.)","Clicks"])
ax.grid(axis="y", alpha=.2)
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

## 4. Country distribution: a segment to investigate

The US is the largest impression market; China contributes an unusually high share of clicks relative to its impression share. Check query and landing-page mix before attributing intent or making localization changes.

```python
countries = pd.DataFrame([
    {"country":"United States","impressions":14100,"clicks":143},
    {"country":"Rest of World","impressions":2600,"clicks":48},
    {"country":"China","impressions":1700,"clicks":159},
    {"country":"United Kingdom","impressions":702,"clicks":14},
    {"country":"Canada","impressions":278,"clicks":14},
    {"country":"Germany","impressions":262,"clicks":4},
    {"country":"France","impressions":168,"clicks":1},
    {"country":"Japan","impressions":93,"clicks":7},
    {"country":"Spain","impressions":50,"clicks":1},
    {"country":"Brazil","impressions":42,"clicks":0},
    {"country":"Italy","impressions":34,"clicks":1},
])
countries["click_share (%)"] = countries.clicks / 392 * 100
countries["impression_share (%)"] = countries.impressions / 20000 * 100  # approximate; Bing rounds
countries["CTR (%)"] = countries.clicks / countries.impressions * 100
countries.sort_values("click_share (%)", ascending=False).head(7)
```

```python
top_countries = countries.sort_values("click_share (%)", ascending=True).tail(7)
ax = top_countries.set_index("country")[["click_share (%)","impression_share (%)"]].plot(
    kind="barh", figsize=(8,5), color=["#7663A6","#D18A38"]
)
ax.set_xlabel("Share of Bing total (%)")
ax.set_title("Clicks and impressions by country")
ax.legend(["Clicks","Impressions (approx.)"])
ax.grid(axis="x", alpha=.2)
plt.tight_layout()
plt.show()
```

- **United States:** 143 clicks (**36.5%**) from about 14.1K impressions (**70.5%**).
- **China:** 159 clicks (**40.6%**) from about 1.7K impressions (**8.5%**), with 9.29% CTR.
- **Rest of World:** 48 clicks (**12.2%**) from about 2.6K impressions (**13.0%**).

Country labels are aggregate search segments, not user identity or language preference.

## 7. Product-growth conclusion: what problem should Dafinchi solve?

### The clearest evidence-backed user task

The leading visible Bing searches are not generic finance terms. They ask about **specific company changes**: tariffs, inventory and China demand, CRE exposure, a forecast/acquisition, and cash runway. Together they total about **2,770 impressions (13.9% of Bing’s rounded 20K total)** and zero clicks. Within those five rows, the topic mix is approximately:

| Search task | Impressions | Share of top-five impressions |
|---|---:|---:|
| Whirlpool / tariffs | ~1,000 | 36.1% |
| A. O. Smith / inventory and China demand | 702 | 25.3% |
| Ameris / CRE risk | 530 | 19.1% |
| Abbott / forecast and acquisition | 295 | 10.6% |
| Axsome / cash runway | 243 | 8.8% |

**What that means in product terms:** the best-supported target task is helping someone understand *what changed at a company and why it matters*. The data supports this research task; it does not identify the searchers’ job titles or prove they are investors.

### There is a small set of content examples to learn from

The visible Broadcom, Altria, and Synopsys pages together had **34 clicks from 365 impressions (9.3% CTR)**, compared with Bing’s **1.96% overall CTR**. These three selected pages account for about **8.7% of Bing clicks from 1.8% of rounded impressions**. This is a useful teardown set, not a causal benchmark: inspect their query match, titles, first-screen answer, and links to supporting evidence, then see what can be reused on similar pages.

### Product decision to test

Pilot a path that makes the research task work end to end:

**Specific company search → concise explanation of the catalyst → transcript passage that supports it → compare peers or explore related topics using Dafinchi.**

That fits the visible queries and Dafinchi’s transcript, comparison, Keyword Scan, and Earnings Chat capabilities. It remains a product hypothesis until Analytics confirms visitors use those features.

### Why this is a growth problem, not just a content task

Bing shows discovery: **392 clicks and about 20K impressions**. But Search Console reports only **1 indexed page and 26,810 not indexed**, and its performance page report showed the homepage. That points to a separate Google distribution constraint. Audit URLs that matter by template and indexing reason; don’t try to index every URL.

The zero-click pages expose another leak between appearing in results and earning a visit. For example, the Whirlpool page had 975 impressions, zero clicks, and average position 1.79. Review the actual search result and page before deciding whether the mismatch is the snippet, intent, or page answer.

### What the current data cannot establish

Search reports do not say whether visitors read a transcript, use a Dafinchi tool, sign up, return, or are automated. Google Analytics was inaccessible. Therefore we can prioritize an acquisition-to-product experiment, but we cannot claim the current audience converts or retains until those events are measured.
