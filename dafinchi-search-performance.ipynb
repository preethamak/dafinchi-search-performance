```python
import matplotlib.pyplot as plt

# Values transcribed from the visible reports. Bing rounds some impression totals.
search_clicks = {"Bing": 392, "Google Search Console": 8}
device_clicks = {"Desktop": 379, "Mobile": 13}
country_clicks = {"China": 159, "United States": 143, "Rest of World": 48, "United Kingdom": 14, "Canada": 14, "Japan": 7, "Germany": 4, "France": 1, "Spain": 1, "Italy": 1, "Brazil": 0}
top_query_impressions = {"Whirlpool / Section 232": 1000, "A. O. Smith / channel inventory": 702, "Ameris Bancorp / CRE concentration": 530, "Abbott / Exact Sciences acquisition": 295, "Axsome / cash runway": 243}

def plot_bar(data, title, ylabel, horizontal=False, color="#3569A8", note=None):
    labels, values = list(data), list(data.values())
    fig, ax = plt.subplots(figsize=(9, 4.8))
    if horizontal:
        y = list(range(len(labels)))
        ax.barh(y, values, color=color, height=.65)
        ax.set_yticks(y, labels)
        ax.invert_yaxis()
        ax.set_xlabel(ylabel)
        ax.grid(axis="x", alpha=.2)
        for i, value in enumerate(values):
            ax.text(value + max(values or [1])*.015, i, f"{value:,}", va="center", fontsize=9)
    else:
        ax.bar(labels, values, color=color, width=.62)
        ax.set_ylabel(ylabel)
        ax.grid(axis="y", alpha=.2)
        for i, value in enumerate(values):
            ax.text(i, value + max(values or [1])*.025, f"{value:,}", ha="center", fontsize=9)
    ax.set_title(title, loc="left", weight="bold", fontsize=13, pad=14)
    ax.set_axisbelow(True)
    for spine in ["top", "right"]:
        ax.spines[spine].set_visible(False)
    if note:
        fig.text(.01, .005, note, fontsize=8, color="#555")
    fig.tight_layout(rect=(0, .04, 1, 1))
    fig
```

```python
plot_bar(search_clicks, "Reported organic search clicks", "Clicks", note="Bing: Aug 15–Nov 14, 2026. Google: Aug 14–Nov 13, 2026.")
```

```python
plot_bar(device_clicks, "Bing search clicks by device", "Clicks", color="#2A8C82")
```

```python
plot_bar(country_clicks, "Bing search clicks by country", "Clicks", horizontal=True, color="#7A61A8", note="Source: Bing Search Performance, Aug 15–Nov 14, 2026.")
```

```python
plot_bar(top_query_impressions, "Bing queries with the most impressions", "Impressions", horizontal=True, color="#D58A3A", note="Whirlpool value is shown as 1.0K by Bing and plotted as approximately 1,000.")
```
