## 3. Device distribution: where acquisition is strongest

Bing desktop drove **96.7% of clicks** from about **89% of impressions**. Mobile drove **3.3% of clicks** from about **11% of impressions**. Mobile CTR was **0.59%**, compared with **2.13%** on desktop; average position was also worse (7.39 vs. 5.44).

If mobile’s current rounded 2.2K impressions had desktop CTR, that would imply about **47 clicks**, versus 13 observed. The ~34-click difference is a scenario to size the gap, not a forecast: mobile query mix and rankings differ.

```python
import pandas as pd
import matplotlib.pyplot as plt

devices = pd.DataFrame([
    {"device":"Desktop", "impressions":17800, "clicks":379, "CTR (%)":2.13, "avg_position":5.44},
    {"device":"Mobile", "impressions":2200, "clicks":13, "CTR (%)":0.59, "avg_position":7.39},
])
devices["impression_share (%)"] = devices["impressions"] / 20000 * 100  # Bing total is rounded
devices["click_share (%)"] = devices["clicks"] / 392 * 100
devices[["device","impressions","clicks","CTR (%)","avg_position","impression_share (%)","click_share (%)"]]
```

```python
ax = devices.set_index("device")[["impression_share (%)","click_share (%)"]].plot(
    kind="bar", figsize=(8,4), color=["#3268A8","#2A8C82"]
)
ax.set_ylabel("Share of Bing total (%)")
ax.set_title("Bing search distribution by device")
ax.legend(["Impressions (approx.)","Clicks"])
ax.grid(axis="y", alpha=.2)
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

## 4. Country distribution: a segment to investigate

The US is the largest impression market; China contributes an unusually high share of clicks relative to its impression share. Check query and landing-page mix before attributing intent or making localization changes.

```python
countries = pd.DataFrame([
    {"country":"United States","impressions":14100,"clicks":143},
    {"country":"Rest of World","impressions":2600,"clicks":48},
    {"country":"China","impressions":1700,"clicks":159},
    {"country":"United Kingdom","impressions":702,"clicks":14},
    {"country":"Canada","impressions":278,"clicks":14},
    {"country":"Germany","impressions":262,"clicks":4},
    {"country":"France","impressions":168,"clicks":1},
    {"country":"Japan","impressions":93,"clicks":7},
    {"country":"Spain","impressions":50,"clicks":1},
    {"country":"Brazil","impressions":42,"clicks":0},
    {"country":"Italy","impressions":34,"clicks":1},
])
countries["click_share (%)"] = countries.clicks / 392 * 100
countries["impression_share (%)"] = countries.impressions / 20000 * 100  # approximate; Bing rounds
countries["CTR (%)"] = countries.clicks / countries.impressions * 100
countries.sort_values("click_share (%)", ascending=False).head(7)
```

```python
top_countries = countries.sort_values("click_share (%)", ascending=True).tail(7)
ax = top_countries.set_index("country")[["click_share (%)","impression_share (%)"]].plot(
    kind="barh", figsize=(8,5), color=["#7663A6","#D18A38"]
)
ax.set_xlabel("Share of Bing total (%)")
ax.set_title("Clicks and impressions by country")
ax.legend(["Clicks","Impressions (approx.)"])
ax.grid(axis="x", alpha=.2)
plt.tight_layout()
plt.show()
```

- **United States:** 143 clicks (**36.5%**) from about 14.1K impressions (**70.5%**).
- **China:** 159 clicks (**40.6%**) from about 1.7K impressions (**8.5%**), with 9.29% CTR.
- **Rest of World:** 48 clicks (**12.2%**) from about 2.6K impressions (**13.0%**).

Country labels are aggregate search segments, not user identity or language preference.

## 7. Recommended growth order

1. **Audit index coverage.** Sample valuable URLs among discovered/crawled-not-indexed pages. Review content uniqueness, canonical, internal links, crawl access, and sitemap scope. Measure indexation and qualified query coverage for target pages.
2. **Improve high-impression snippets and page openings.** Start with Whirlpool/Section 232, A. O. Smith, and Ameris. Match the title, description, and first screen to the specific catalyst query. Measure query/page CTR.
3. **Build catalyst briefs that lead into the product.** Put the key metric/change and transcript source near the top, then link naturally to comparison, Keyword Scan, and Earnings Chat. Measure onward product actions from organic landings.
4. **Review mobile discovery and reading.** Compare snippets and first viewport on high-impression pages. Mobile CTR is 0.59% vs 2.13% desktop; its position is also lower.
5. **Investigate China’s query mix.** Check which queries and pages account for 159 clicks and 9.29% CTR before planning country-specific changes.
6. **Instrument the funnel.** When GA4 is accessible, measure organic landing → transcript read → product action → signup/return. Search reports do not show sessions, conversions, or retention.
